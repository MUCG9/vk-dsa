## Узел бинарного дерева

Класс `TreeNode` хранит значение узла и ссылки на его левого и правого потомков.

```python
class TreeNode:
    def __init__(self, data: int, left=None, right=None):
        self.data = data
        self.left = left
        self.right = right
```

## 1. Восстановление бинарного дерева из массива

Дан массив, представляющий бинарное дерево. Необходимо восстановить дерево. Значение `None` обозначает отсутствующий узел.

Корень находится по индексу `0`. Для элемента с индексом `i` левый потомок находится по индексу `2 * i + 1`, а правый — по индексу `2 * i + 2`. Функция рекурсивно создаёт текущий узел и строит оба его поддерева.

Сложность: время `O(n)`, дополнительная память `O(h)` для стека рекурсии, где `h` — высота дерева.

```python
def build_tree(values: list, i: int = 0):
    if i >= len(values) or values[i] is None:
        return None

    root = TreeNode(values[i])
    root.left = build_tree(values, 2 * i + 1)
    root.right = build_tree(values, 2 * i + 2)
    return root
```

## 2. Симметричное бинарное дерево

Дано бинарное дерево. Необходимо определить, является ли оно симметричным относительно своей середины.

Два поддерева зеркальны, если значения их корней равны, левое поддерево первого зеркально правому поддереву второго, а правое поддерево первого — левому поддереву второго. Пары узлов сравниваются именно в таком порядке на каждом уровне.

Сложность: время `O(n)`, дополнительная память `O(h)` для стека рекурсии.

```python
def is_symmetric(root) -> bool:
    def is_mirror(left, right) -> bool:
        if left is None and right is None:
            return True
        if left is None or right is None:
            return False
        if left.data != right.data:
            return False

        return (
            is_mirror(left.left, right.right)
            and is_mirror(left.right, right.left)
        )

    return root is None or is_mirror(root.left, root.right)
```

## 3. Поиск минимальной глубины бинарного дерева

Дано бинарное дерево. Необходимо найти его минимальную глубину (расстояние от корня до ближайшего листа).

Поиск в ширину обходит дерево по уровням. Первый встреченный лист имеет наименьшую возможную глубину, поэтому после его обнаружения ответ можно сразу вернуть. В очереди вместе с узлом хранится его глубина.

Сложность: время `O(n)`, дополнительная память `O(w)`, где `w` — максимальная ширина дерева.

```python
from collections import deque


def min_depth(root) -> int:
    if root is None:
        return 0

    queue = deque([(root, 1)])

    while queue:
        node, depth = queue.popleft()

        if node.left is None and node.right is None:
            return depth

        if node.left is not None:
            queue.append((node.left, depth + 1))
        if node.right is not None:
            queue.append((node.right, depth + 1))
```

## 4. Произведение минимального и максимального элементов BST

Дано бинарное дерево поиска (BST), представленное в виде массива. Необходимо найти произведение минимального и максимального значений.

После восстановления дерева минимальный элемент находится в самом левом узле, а максимальный — в самом правом. На каждом шаге выбирается только один потомок, поэтому оба поиска зависят от высоты дерева.

Сложность: построение дерева занимает `O(n)`, поиск крайних элементов — `O(h)`. Дополнительная память — `O(h)` для построения дерева.

```python
def min_max_product(values: list) -> int:
    root = build_tree(values)

    minimum = root
    while minimum.left is not None:
        minimum = minimum.left

    maximum = root
    while maximum.right is not None:
        maximum = maximum.right

    return minimum.data * maximum.data
```

## 5. Сравнение двух бинарных деревьев

Даны два бинарных дерева. Необходимо определить, являются ли они идентичными (по структуре и значениям узлов).

Если оба текущих узла отсутствуют, поддеревья равны. Если отсутствует только один узел или значения различаются, деревья не равны. В остальных случаях рекурсивно сравниваются соответствующие левые и правые поддеревья.

Сложность: время `O(n)`, дополнительная память `O(h)` для стека рекурсии.

```python
def is_same_tree(first, second) -> bool:
    if first is None and second is None:
        return True
    if first is None or second is None:
        return False
    if first.data != second.data:
        return False

    return (
        is_same_tree(first.left, second.left)
        and is_same_tree(first.right, second.right)
    )
```
