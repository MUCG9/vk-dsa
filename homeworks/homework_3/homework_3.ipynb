## 1. Найти корень числа (ближайшее целое)

Необходимо найти целую часть квадратного корня неотрицательного числа без использования встроенной функции извлечения корня.

Бинарный поиск выполняется на отрезке от `0` до `number`. Если квадрат середины больше исходного числа, правая граница сдвигается влево. Если квадрат меньше, левая граница сдвигается вправо. Когда точного корня нет, после окончания поиска `right` указывает на наибольшее целое число, квадрат которого не превышает `number`.

Сложность: время `O(log n)`, дополнительная память `O(1)`.

```python
def integer_sqrt(number: int) -> int:
    if number < 0:
        raise ValueError("Число должно быть неотрицательным")

    left = 0
    right = number

    while left <= right:
        middle = (left + right) // 2
        square = middle * middle

        if square == number:
            return middle
        if square < number:
            left = middle + 1
        else:
            right = middle - 1

    return right
```

## 2. Очень лёгкая задача (два ксерокса)

Есть два ксерокса: первый делает копию за `x` минут, второй - за `y` минут. Необходимо найти минимальное время, за которое можно сделать `copies` копий документа.

Первую копию делает самый быстрый ксерокс за `min(x, y)` минут. После этого можно использовать оба устройства одновременно. Для оставшихся `copies - 1` копий выполняется бинарный поиск по времени. За `middle` минут ксероксы успевают сделать `middle // x + middle // y` копий.

Сложность: время `O(log(copies · max(x, y)))`, дополнительная память `O(1)`.

```python
def copy_time(copies: int, x: int, y: int) -> int:
    if copies <= 0:
        return 0
    if x <= 0 or y <= 0:
        raise ValueError("Время копирования должно быть положительным")

    first_copy_time = min(x, y)
    copies_left = copies - 1
    left = 0
    right = copies_left * max(x, y)

    while left < right:
        middle = (left + right) // 2
        made = middle // x + middle // y

        if made >= copies_left:
            right = middle
        else:
            left = middle + 1

    return first_copy_time + left
```

## 3. Накормить животных

Массив `animals` содержит потребности животных в еде, а массив `food` - размеры привезённых порций. Одна порция может накормить только одно животное и не делится. Необходимо накормить максимальное количество животных.

Оба массива сортируются по возрастанию. Каждая порция сравнивается с самым неприхотливым из ещё не накормленных животных. Если порции достаточно, животное считается накормленным. Такая стратегия не расходует большую порцию на животное, которое можно накормить меньшей порцией.

Сложность: время `O(n log n + m log m)`, где `n` - число животных, `m` - число порций. Дополнительная память определяется реализацией сортировки Python.

```python
def feed_animals(animals: list[int], food: list[int]) -> int:
    if not animals or not food:
        return 0

    animals.sort()
    food.sort()
    fed = 0

    for portion in food:
        if portion >= animals[fed]:
            fed += 1

        if fed == len(animals):
            break

    return fed
```

## 4. Найти разницу между двумя строками

Строка `b` получена из строки `a` перемешиванием символов и добавлением одной новой буквы. Необходимо вернуть добавленную букву без сортировки строк.

Сначала в словаре подсчитываются все символы строки `b`. Затем для каждого символа строки `a` соответствующий счётчик уменьшается. После этого положительный счётчик остаётся только у добавленного символа.

Сложность: время `O(n)`, дополнительная память `O(k)`, где `k` - число различных символов.

```python
def extra_letter(a: str, b: str) -> str:
    counts = {}

    for letter in b:
        counts[letter] = counts.get(letter, 0) + 1

    for letter in a:
        counts[letter] -= 1

    for letter, count in counts.items():
        if count > 0:
            return letter

    return ""
```

## 5. Сумма двух элементов массива

Дан неотсортированный массив целых чисел и число `target`. Необходимо вернуть индексы двух разных элементов, сумма которых равна `target`. Если подходящей пары нет, возвращается пустой список.

Во время одного прохода словарь `seen` хранит уже просмотренные значения и их индексы. Для текущего элемента вычисляется дополнение `target - value`. Если оно уже находится в словаре, пара найдена.

Сложность: время `O(n)`, дополнительная память `O(n)`.

```python
def two_sum(nums: list[int], target: int) -> list[int]:
    seen = {}

    for index, value in enumerate(nums):
        complement = target - value

        if complement in seen:
            return [seen[complement], index]

        seen[value] = index

    return []
```

## 6. Массив анаграмм

Дан массив строк `strs`. Необходимо сгруппировать слова, являющиеся анаграммами. Порядок групп не имеет значения.

Для каждого слова строится таблица частот символов. Набор пар `(символ, количество)` используется как хешируемый ключ словаря. У анаграмм совпадают и символы, и количество их повторений, поэтому они попадают в одну группу.

Сложность: время `O(S)`, где `S` - суммарное число символов во всех строках. Дополнительная память `O(S)`.

```python
def group_anagrams(strs: list[str]) -> list[list[str]]:
    groups = {}

    for word in strs:
        counts = {}
        for letter in word:
            counts[letter] = counts.get(letter, 0) + 1

        key = frozenset(counts.items())
        groups.setdefault(key, []).append(word)

    return list(groups.values())
```
