## Узел односвязного списка

Класс `ListNode` задаёт узел односвязного списка. Каждый узел хранит значение и ссылку на следующий элемент.

```python
class ListNode:
    def __init__(self, value: int, next_node=None):
        self.value = value
        self.next = next_node
```

## 1. Разворот односвязного списка

Необходимо написать функцию, которая принимает односвязный список и разворачивает его in-place, изменяя ссылки между узлами.

На каждом шаге сохраняется следующий узел, после чего ссылка текущего узла направляется назад. Затем оба указателя сдвигаются вперёд.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def reverse_linked_list(head):
    previous = None
    current = head

    while current is not None:
        following = current.next
        current.next = previous
        previous = current
        current = following

    return previous
```

## 2. Середина списка

Дан односвязный список. Необходимо найти его середину за `O(n)` без дополнительных аллокаций.

Указатель `slow` проходит по одному узлу, а `fast` — по два. Когда быстрый указатель доходит до конца, медленный находится в середине. Для чётной длины получается вторая середина.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def middle(head):
    slow = fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    return slow
```

## 3. Удаление элемента

Необходимо написать функцию, которая принимает односвязный список и целое число `val`. Из списка нужно удалить узлы, значение которых равно `val`.

Фиктивный узел перед головой позволяет не выделять удаление первого элемента в отдельный случай. При совпадении предыдущий узел связывается со следующим. Удаляются все узлы со значением `val`.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def remove_elements(head, val: int):
    dummy = ListNode(0, head)
    previous = dummy
    current = head

    while current is not None:
        if current.value == val:
            previous.next = current.next
        else:
            previous = current
        current = current.next

    return dummy.next
```

## 4. Является ли одна строка исходной для другой

В исходную строку добавили некоторое количество символов. Необходимо определить, была ли строка `a` исходной для строки `b`.

Задача сводится к проверке подпоследовательности. Строка `a` является исходной для `b`, если все её символы встречаются в `b` в том же порядке. Указатель по `a` сдвигается только при совпадении.

Сложность: время `O(len(b))`, дополнительная память `O(1)`.

```python
def is_subsequence(a: str, b: str) -> bool:
    index = 0

    for element in b:
        if index < len(a) and a[index] == element:
            index += 1

    return index == len(a)
```

## 5. Поиск пары с заданной суммой

Дан отсортированный массив чисел и целевое число `target`. Необходимо определить, существуют ли два разных элемента, сумма которых равна `target`.

В отсортированном массиве ставим указатели на края. Если сумма мала, сдвигаем левый указатель вправо; если велика — правый влево. Условие `left < right` гарантирует, что используются два разных элемента.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def pair_sum(nums: list[int], target: int) -> bool:
    left = 0
    right = len(nums) - 1

    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return True
        if current_sum < target:
            left += 1
        else:
            right -= 1

    return False
```

## 6. Палиндром

Необходимо написать функцию, которая принимает строку и возвращает `True`, если строка является палиндромом, и `False` в противном случае.

Левый и правый символы сравниваются попарно, указатели сходятся к центру. При первом несовпадении можно сразу вернуть `False`.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def is_palindrome(word: str) -> bool:
    left = 0
    right = len(word) - 1

    while left < right:
        if word[left] != word[right]:
            return False
        left += 1
        right -= 1

    return True
```

## 7. Удаление дубликатов

Дан отсортированный массив. Необходимо удалить дубликаты in-place так, чтобы каждый элемент встречался только один раз.

Медленный указатель отмечает позицию для следующего уникального элемента, быстрый просматривает массив. Так как массив отсортирован, достаточно сравнивать новый элемент с последним записанным. После прохода лишний хвост удаляется.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def remove_duplicates(nums: list[int]) -> int:
    if not nums:
        return 0

    write = 1
    for read in range(1, len(nums)):
        if nums[read] != nums[write - 1]:
            nums[write] = nums[read]
            write += 1

    del nums[write:]
    return write
```

## 8. Слияние двух отсортированных списков

Даны два отсортированных односвязных списка. Необходимо объединить их в один отсортированный список, изменяя только ссылки между существующими узлами.

Сначала выбирается голова результата, после меньший из двух текущих узлов присоединяется к хвосту. Новые узлы не создаются, меняются только ссылки существующих узлов.

Сложность: время `O(n + m)`, дополнительная память `O(1)`.

```python
def merge_sorted(first, second):
    if first is None:
        return second
    if second is None:
        return first

    if first.value <= second.value:
        head = tail = first
        first = first.next
    else:
        head = tail = second
        second = second.next

    while first is not None and second is not None:
        if first.value <= second.value:
            tail.next = first
            first = first.next
        else:
            tail.next = second
            second = second.next
        tail = tail.next

    tail.next = first if first is not None else second
    return head
```
