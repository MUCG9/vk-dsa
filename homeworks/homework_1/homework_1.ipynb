## 1. Циклический сдвиг массива

Дан массив целых чисел. Необходимо повернуть (сдвинуть) справа налево часть массива, которая указана параметром k. Решение должно работать за линейное время без дополнительных аллокаций.

Сначала разворачивается весь массив, затем первые `k` элементов и оставшаяся часть. Если `k` больше длины массива, используется остаток от деления.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def reverse(nums: list[int], left: int, right: int) -> None:
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1


def rotate(nums: list[int], k: int) -> None:
    if not nums:
        return

    k %= len(nums)
    if k == 0:
        return

    reverse(nums, 0, len(nums) - 1)
    reverse(nums, 0, k - 1)
    reverse(nums, k, len(nums) - 1)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7]
rotate(numbers, 3)
check("Test 1 (циклический сдвиг)", numbers == [5, 6, 7, 1, 2, 3, 4])

numbers = [1, 2, 3]
rotate(numbers, 7)
check("Test 2 (k больше длины массива)", numbers == [3, 1, 2])

numbers = []
rotate(numbers, 5)
check("Test 3 (пустой массив)", numbers == [])
```

## 2. Слияние двух отсортированных массивов

Дано два отсортированных массива. Необходимо написать функцию, которая объединит эти два массива в один отсортированный.

Два указателя идут по входным массивам. В результат добавляется меньший из текущих элементов. После окончания одного массива дописывается остаток второго.

Сложность: время `O(n + m)`, память `O(n + m)` для результата.

```python
def merge_sorted(A: list[int], B: list[int]) -> list[int]:
    merged = []
    left = right = 0

    while left < len(A) and right < len(B):
        if A[left] <= B[right]:
            merged.append(A[left])
            left += 1
        else:
            merged.append(B[right])
            right += 1

    merged.extend(A[left:])
    merged.extend(B[right:])
    return merged
```

```python
result = merge_sorted([1, 3, 5, 5], [2, 4, 5, 6])
check("Test 1 (слияние массивов)", result == [1, 2, 3, 4, 5, 5, 5, 6])
check("Test 2 (слияние с пустым массивом)", merge_sorted([], [1, 2]) == [1, 2])
```

## 3. Слияние двух отсортированных массивов (без дополнительных аллокаций)

Дано два отсортированных массива. Первый массив имеет размер результирующего массива, а его конец заполнен нулями. Объединить массивы необходимо без создания третьего массива.

В конце первого массива заранее оставлено место. Заполнение идёт справа налево, поэтому ещё не прочитанные элементы первого массива не перезаписываются.

Сложность: время `O(n + m)`, дополнительная память `O(1)`.

```python
def merge_sorted_arrays(
    first: list[int], second: list[int]) -> None:
    pointer1 = len(first) - len(second) - 1
    pointer2 = len(second) - 1
    pointer3 = len(first) - 1

    while pointer2 >= 0:
        if pointer1 >= 0 and first[pointer1] > second[pointer2]:
            first[pointer3] = first[pointer1]
            pointer1 -= 1
        else:
            first[pointer3] = second[pointer2]
            pointer2 -= 1
        pointer3 -= 1
```

```python
first = [1, 2, 3, 0, 0, 0]
merge_sorted_arrays(first, [2, 5, 6])
check("Test 1", first == [1, 2, 2, 3, 5, 6])

first = [-1, 0, 0, 0]
merge_sorted_arrays(first, [0, 2])
check("Test 2", first == [-1, 0, 0, 2])
```

## 4. Сортировка массива из 0 и 1

Дан массив, содержащий только 0 и 1. Напишите функцию, которая сортирует его так, чтобы все нули оказались в начале, а единицы — в конце. Решение должно быть in-place.

Указатель `zero_position` хранит место для следующего нуля. При встрече нуля он меняется местами с элементом на этой позиции.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def sort_binary_array(nums: list[int]) -> None:
    zero_position = 0

    for current, value in enumerate(nums):
        if value == 0:
            nums[zero_position], nums[current] = nums[current], nums[zero_position]
            zero_position += 1
```

```python
numbers = [1, 0, 1, 1, 0, 0, 1]
sort_binary_array(numbers)
check("Test 1", numbers == [0, 0, 0, 1, 1, 1, 1])
```

## 5. Флаг Нидерландов

Дан массив, состоящий из 0, 1 и 2. Необходимо отсортировать его за линейное время. Реализуйте алгоритм трёх указателей (Dutch National Flag).

Поддерживаются четыре области: нули `[0, low)`, единицы `[low, mid)`, непроверенная часть `[mid, high]` и двойки `(high, n)`. При обработке двойки `mid` не увеличивается, поскольку элемент с позиции `high` ещё нужно проверить.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def sort_colors(nums: list[int]) -> None:
    low = mid = 0
    high = len(nums) - 1

    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        elif nums[mid] == 2:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
```

```python
numbers = [2, 0, 2, 1, 1, 0]
sort_colors(numbers)
check("флаг Нидерландов", numbers == [0, 0, 1, 1, 2, 2])
```

## 6. Чётные числа вперёд

Дан неотсортированный массив целых чисел. Необходимо перенести в начало массива все чётные числа, сохраняя их относительный порядок.

Чётные числа обрабатываются слева направо и по очереди ставятся в начало массива, поэтому их взаимный порядок сохраняется. Порядок нечётных элементов может измениться.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def evens_first(nums: list[int]) -> None:
    even_position = 0

    for current in range(len(nums)):
        if nums[current] % 2 == 0:
            nums[even_position], nums[current] = nums[current], nums[even_position]
            even_position += 1
```

```python
numbers = [3, 2, 4, 1, 11, 8, 9]
evens_first(numbers)
check("Test 1", numbers == [2, 4, 8, 1, 11, 3, 9])
```

## 7. Нули в конец

Дан массив оценок, среди которых есть 0. Необходимо перенести все нули в конец массива. Порядок остальных элементов должен сохраняться.

Сначала все ненулевые элементы записываются в начало в исходном порядке. Оставшиеся позиции заполняются нулями.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def move_zeroes(nums: list[int]) -> None:
    pointer = 0

    for value in nums:
        if value != 0:
            nums[pointer] = value
            pointer += 1

    while pointer < len(nums):
        nums[pointer] = 0
        pointer += 1
```

```python
numbers = [0, 33, 57, 88, 60, 0, 0, 80, 99]
move_zeroes(numbers)
check("Test 1", numbers == [33, 57, 88, 60, 80, 99, 0, 0, 0])
```
