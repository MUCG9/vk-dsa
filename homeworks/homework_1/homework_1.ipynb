## 1. Циклический сдвиг массива

Необходимо повернуть (сдвинуть) справа налево часть массива, которая указана параметром k.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def reverse(nums: list[int], left: int, right: int) -> None:
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1


def rotate(nums: list[int], k: int) -> None:
    if not nums:
        return

    k %= len(nums)
    if k == 0:
        return

    reverse(nums, 0, len(nums) - 1)
    reverse(nums, 0, k - 1)
    reverse(nums, k, len(nums) - 1)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7]
rotate(numbers, 3)
check("Test 1 (циклический сдвиг)", numbers == [5, 6, 7, 1, 2, 3, 4])

numbers = [1, 2, 3]
rotate(numbers, 7)
check("Test 2 (k больше длины массива)", numbers == [3, 1, 2])

numbers = []
rotate(numbers, 5)
check("Test 3 (пустой массив)", numbers == [])
```

## 2. Слияние двух отсортированных массивов

Реализовать алгоритм слияния двух остортированных массива, используя два указателя.

Сложность: время `O(n + m)`, память `O(n + m)` для результата.

```python
def merge_sorted(A: list[int], B: list[int]) -> list[int]:
    merged = []
    left = right = 0

    while left < len(A) and right < len(B):
        if A[left] <= B[right]:
            merged.append(A[left])
            left += 1
        else:
            merged.append(B[right])
            right += 1

    merged.extend(A[left:])
    merged.extend(B[right:])
    return merged
```

```python
result = merge_sorted([1, 3, 5, 5], [2, 4, 5, 6])
check("Test 1 (слияние массивов)", result == [1, 2, 3, 4, 5, 5, 5, 6])
check("Test 2 (слияние с пустым массивом)", merge_sorted([], [1, 2]) == [1, 2])
```

## 3. Слияние двух отсортированных массивов (без дополнительных аллокаций)

Напишите функцию, которая объединяет массивы, заполняя первый массив с конца. Первый массив имеет размер результирующего массива, а его конец заполнен нулями.

Сложность: время `O(n + m)`, дополнительная память `O(1)`.

```python
def merge_sorted_(
    first: list[int], first_count: int, second: list[int], second_count: int
) -> None:
    left = first_count - 1
    right = second_count - 1
    write = first_count + second_count - 1

    while right >= 0:
        if left >= 0 and first[left] > second[right]:
            first[write] = first[left]
            left -= 1
        else:
            first[write] = second[right]
            right -= 1
        write -= 1
```

## 4. Сортировка массива из 0 и 1

Указатель `zero_position` хранит место для следующего нуля. При встрече нуля он меняется местами с элементом на этой позиции.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def sort_binary_array(nums: list[int]) -> None:
    zero_position = 0

    for current, value in enumerate(nums):
        if value not in (0, 1):
            raise ValueError("Массив должен содержать только 0 и 1")
        if value == 0:
            nums[zero_position], nums[current] = nums[current], nums[zero_position]
            zero_position += 1
```

```python
values = [1, 0, 1, 1, 0, 0, 1]
sort_binary_array(values)
check("бинарный массив", values == [0, 0, 0, 1, 1, 1, 1])
```

## 5. Флаг Нидерландов

Поддерживаются четыре области: нули `[0, low)`, единицы `[low, mid)`, непроверенная часть `[mid, high]` и двойки `(high, n)`. При обработке двойки `mid` не увеличивается, потому что элемент, пришедший с позиции `high`, ещё нужно проверить.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def sort_colors(nums: list[int]) -> None:
    low = mid = 0
    high = len(nums) - 1

    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        elif nums[mid] == 2:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
        else:
            raise ValueError("Массив должен содержать только 0, 1 и 2")
```

```python
values = [2, 0, 2, 1, 1, 0]
sort_colors(values)
check("флаг Нидерландов", values == [0, 0, 1, 1, 2, 2])
```

## 6. Чётные числа вперёд

Чётные числа обрабатываются слева направо и по очереди ставятся в начало массива. Поэтому их взаимный порядок сохраняется. Порядок нечётных элементов, как и в примере задания, может измениться.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def move_evens_to_front(nums: list[int]) -> None:
    even_position = 0

    for current in range(len(nums)):
        if nums[current] % 2 == 0:
            nums[even_position], nums[current] = nums[current], nums[even_position]
            even_position += 1
```

```python
values = [3, 2, 4, 1, 11, 8, 9]
move_evens_to_front(values)
check("чётные числа впереди", values == [2, 4, 8, 1, 11, 3, 9])
```

## 7. Нули в конец

Сначала все ненулевые элементы записываются в начало в исходном порядке. Оставшиеся позиции заполняются нулями.

Сложность: время `O(n)`, дополнительная память `O(1)`.

```python
def move_zeroes(nums: list[int]) -> None:
    write = 0

    for value in nums:
        if value != 0:
            nums[write] = value
            write += 1

    while write < len(nums):
        nums[write] = 0
        write += 1
```

```python
values = [0, 33, 57, 88, 60, 0, 0, 80, 99]
move_zeroes(values)
check("нули в конце", values == [33, 57, 88, 60, 80, 99, 0, 0, 0])
print("Все тесты домашнего задания № 1 пройдены.")
```
